```python
import os, shutil, glob

os.makedirs("/content", exist_ok=True)

for pattern in ["co2-emissions-per-capita.csv", "primary-energy-cons.csv"]:
    if not os.path.exists(f"/content/{pattern}"):
        matches = glob.glob(pattern) + glob.glob(f"data/{pattern}") + glob.glob(f"**/{pattern}", recursive=True)
        if matches:
            shutil.copy(matches[0], f"/content/{pattern}")
```

```python
import pandas as pd

df_co2 = pd.read_csv("/content/co2-emissions-per-capita.csv")
df_primary_energy = pd.read_csv("/content/primary-energy-cons.csv")

pl_co2 = df_co2[df_co2["Entity"] == "Poland"][["Entity","Year","Annual CO₂ emissions (per capita)"]]
pl_en = df_primary_energy[df_primary_energy["Entity"] == "Poland"][["Entity","Year","Primary energy consumption (TWh)"]]
pl = pl_en.merge(pl_co2, on=["Entity","Year"], how="inner").sort_values("Year")
pl.to_csv("polska_energia_co2.csv", index=False)

eu_name = "European Union (27)" if (df_co2["Entity"] == "European Union (27)").any() else "European Union (28)"
eu_co2 = df_co2[df_co2["Entity"] == eu_name][["Entity","Year","Annual CO₂ emissions (per capita)"]]
eu_en = df_primary_energy[df_primary_energy["Entity"] == eu_name][["Entity","Year","Primary energy consumption (TWh)"]]
eu = eu_en.merge(eu_co2, on=["Entity","Year"], how="inner").sort_values("Year")

us_co2 = df_co2[df_co2["Entity"] == "United States"][["Entity","Year","Annual CO₂ emissions (per capita)"]]
us_en = df_primary_energy[df_primary_energy["Entity"] == "United States"][["Entity","Year","Primary energy consumption (TWh)"]]
us = us_en.merge(us_co2, on=["Entity","Year"], how="inner").sort_values("Year")

with pd.ExcelWriter("energia_co2.xlsx") as writer:
    eu.to_excel(writer, sheet_name="UE", index=False)
    us.to_excel(writer, sheet_name="USA", index=False)

pl.head()
```
